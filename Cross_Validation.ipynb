```python
# Standard k-fold Cross Validation
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import KFold
kfold = KFold(n_splits = 8)

iris = load_iris()

LR = LogisticRegression(max_iter=10000)
scores = cross_val_score(LR, iris.data, iris.target, cv=kfold)
print(scores)
```

```python
# Leave One Out
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import LeaveOneOut
LOO = LeaveOneOut()

iris = load_iris()

LR = LogisticRegression(max_iter=10000)
scores = cross_val_score(LR, iris.data, iris.target, cv=LOO)
print(scores)
```

```python
# Shuffle Split Method
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import ShuffleSplit
SS = ShuffleSplit(test_size = 0.5, n_splits=5)

iris = load_iris()

LR = LogisticRegression(max_iter=10000)
scores = cross_val_score(LR, iris.data, iris.target, cv=SS)
print(scores)
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score

cancer = load_breast_cancer()
X_train,X_test,y_train,y_test=train_test_split(cancer.data, cancer.target)
LR = LogisticRegression(max_iter=10000).fit(X_train, y_train)

print("Training accuracy",LR.score(X_train, y_train))
print("Test accuracy", LR.score(X_test,y_test))
p=LR.predict(X_test)
print(accuracy_score(y_test,p))
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, p)
```
