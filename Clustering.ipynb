```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
iris = load_iris()

S=StandardScaler()
iris = S.fit_transform(iris.data)
model = KMeans(n_clusters = 3, init = "k-means++", max_iter = 300, n_init = 10, random_state = 0).fit(iris)

import matplotlib.pyplot as plt
plt.scatter(iris[:,0],iris[:,1], cmap = "rainbow", c = model.labels_)
plt.show()
```

```python
#Agglomeratrive Clustering
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
iris = load_iris()

S=StandardScaler()
iris = S.fit_transform(iris.data)
model = AgglomerativeClustering().fit(iris)

import matplotlib.pyplot as plt
plt.scatter(iris[:,0],iris[:,1], cmap = "rainbow", c = model.labels_)
plt.show()
```

```python
#DBSCAN - DENSITY BASED SCAN
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
iris = load_iris()

S=StandardScaler()
iris = S.fit_transform(iris.data)
model = DBSCAN().fit(iris)

import matplotlib.pyplot as plt
plt.scatter(iris[:,0],iris[:,1], cmap = "rainbow", c = model.labels_)
plt.show()
```
